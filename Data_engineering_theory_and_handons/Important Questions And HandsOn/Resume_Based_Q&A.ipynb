### **The Auto Loader (cloudFiles) Code**

Here is the exact PySpark code you would write to replace the old Azure SQL watermark method For **Incremental Load**.

---

```python
# 1. Define your Azure ADLS paths
source_data_path = "abfss://raw@yourdatalake.dfs.core.windows.net/cpg_data/"
checkpoint_path = "abfss://checkpoints@yourdatalake.dfs.core.windows.net/cpg_silver_load/"
target_table_name = "hive_metastore.silver.cpg_transactions"

# 2. READ: Ingest new files incrementally using Auto Loader
incremental_df = (
    spark.readStream
    .format("cloudFiles") # This is the keyword that activates Auto Loader
    .option("cloudFiles.format", "json") # Format of the raw data
    .option("cloudFiles.schemaLocation", checkpoint_path) # Handles schema drift automatically
    .load(source_data_path)
)

# 3. TRANSFORM: Add basic Medallion architecture metadata
from pyspark.sql.functions import current_timestamp
transformed_df = incremental_df.withColumn("ingest_timestamp", current_timestamp())

# 4. WRITE: Save to the Delta Lake table
(
    transformed_df.writeStream
    .format("delta")
    .option("checkpointLocation", checkpoint_path) # Stores streaming progress metadata; replaces manual watermarking
    .trigger(availableNow=True) # Runs as an incremental batch job for cost efficiency
    .toTable(target_table_name)
)

```

### **How to explain this code in the interview (Your Talking Points):**

If the AI asks how you implemented it, hit these three technical points:

1. **The `cloudFiles` Format:** Instead of writing complex logic to identify new files, I simply set the read stream format to `cloudFiles`. This invokes Databricks Auto Loader, which natively detects new files landing in the Azure Data Lake.
2. **The Checkpoint Directory:** To replace the legacy SQL watermark table, I utilized a **Checkpoint Directory** (`checkpointLocation`). Databricks automatically stores lightweight metadata here to remember exactly which files it has already processed, ensuring zero duplicate processing.
3. **The `availableNow=True` Trigger (Crucial for cost savings):** Since Structured Streaming can be expensive if left running 24/7, I used the `.trigger(availableNow=True)` method. This turns the stream into an incremental batch job. The cluster spins up, processes only the new files tracked by the checkpoint, writes them to the Delta table, and immediately shuts down to save compute costs.

### Azure Data Engineer — Round 1 Interview Prep

#### 1. Azure Data Factory (ADF)

**Q: What is ADF and what are its core components?**

ADF is a cloud-based ETL/ELT and data integration service. 

- **Core components:**

    - **Pipeline** – logical grouping of activities that perform a task
    - **Activity** – a single step (copy, data flow, stored procedure, lookup, etc.)
    - **Dataset** – points to the data you want to use (a table, file, folder)
    - **Linked Service** – connection string / credentials to a data store or compute (like a connection object)
    - **Trigger** – decides when a pipeline runs (schedule, tumbling window, event-based, manual)
    - **Integration Runtime (IR)** – compute infrastructure that executes activities (Azure IR, Self-hosted IR, Azure-SSIS IR)

**Q: Difference between Mapping Data Flow and Wrangling Data Flow?**

- Mapping Data Flow: visual, Spark-based transformation designed by data engineers, compiled and run on Spark clusters ADF manages.
- Wrangling Data Flow: Power Query-based, more for self-service/analysts doing exploratory data prep.

**Q: How do you handle incremental loads in ADF?**
- Use a **watermark column** (e.g., `LastModifiedDate` or an incrementing ID) stored in a control table.
- Lookup activity reads the last watermark value → Copy activity pulls only rows greater than that value → Stored procedure/Set Variable activity updates the watermark after successful load.
- For file-based sources, use **tumbling window triggers** with folder/date partitioning, or change data capture (CDC) if the source supports it.

**Q: What's a Self-hosted Integration Runtime and when do you need it?**

Needed when connecting to on-premises data sources or data behind a private network/VNet that ADF's Azure IR can't reach directly. It's an agent installed on a machine inside that network.

**Q: How do you handle pipeline failures / add error handling?**

- Use "Add activity on failure" paths, `Try-Catch`-like patterns with conditional activities (`If Condition`), retry policies on activities, and `Web`/`Alert` activities or Logic Apps to notify on failure. Log failures to a control/audit table.

**Q: ADF vs Azure Databricks — when do you use which?**

- ADF orchestrates and moves data (ELT/ETL orchestration); Databricks is for heavy, complex transformations at scale using Spark. 
- Common pattern: ADF triggers/orchestrates, Databricks notebook does the transformation.

#### 2. Azure Databricks / PySpark

**Q: What is Databricks and why use it over plain Spark?**

Managed Spark platform with collaborative notebooks, auto-scaling clusters, Delta Lake, job scheduling, and integration with Azure security (AAD, Key Vault, ADLS).

**Q: What is Delta Lake and why does it matter?**

A storage layer on top of Parquet that adds ACID transactions, schema enforcement/evolution, time travel, and efficient upserts (`MERGE`). Solves the "data lake reliability" problem (no dirty reads, no partial writes).

**Q: How do you do an upsert (SCD Type 1) in Databricks?**
```python
from delta.tables import DeltaTable

target = DeltaTable.forPath(spark, "/mnt/delta/target")
target.alias("t").merge(
    source_df.alias("s"),
    "t.id = s.id"
).whenMatchedUpdateAll() \
 .whenNotMatchedInsertAll() \
 .execute()
```

**Q: Difference between `repartition()` and `coalesce()`?**
- `repartition(n)` – full shuffle, can increase or decrease partitions, evens out data.
- `coalesce(n)` – no full shuffle, only reduces partitions, cheaper but can leave data skewed.

**Q: What causes data skew and how do you fix it?**

- Skew = uneven distribution of data across partitions (e.g., one key has way more rows), causing some tasks to run much longer. 
- Fixes: salting the key, broadcasting the smaller table in a join (`broadcast()`), repartitioning on a better key, using AQE (Adaptive Query Execution).

**Q: `.cache()` vs `.persist()`?**

- `cache()` stores in memory only (shortcut for `persist(MEMORY_AND_DISK)` in most versions); 
- `persist()` lets you choose storage level (memory, disk, memory+disk, serialized, etc.). Use when a DataFrame is reused multiple times downstream.

**Q: Common PySpark transformation snippet — read, transform, write**
```python
df = spark.read.format("parquet").load("/mnt/raw/sales")

result = (df.filter(df.amount > 0)
            .withColumn("year", year("order_date"))
            .groupBy("year", "region")
            .agg(sum("amount").alias("total_sales")))

result.write.format("delta").mode("overwrite").partitionBy("year").save("/mnt/curated/sales_summary")
```

**Q: Narrow vs wide transformations?**

- Narrow = no shuffle needed, each output partition depends on one input partition (`filter`, `map`, `withColumn`). 
- Wide = requires shuffle across partitions (`groupBy`, `join`, `distinct`, `orderBy`).

#### 5. SQL Fundamentals

**Q: Types of joins**
- **INNER JOIN** – only matching rows in both tables
- **LEFT JOIN** – all rows from left + matches from right (NULLs if no match)
- **RIGHT JOIN** – opposite of left
- **FULL OUTER JOIN** – all rows from both, matched where possible
- **CROSS JOIN** – Cartesian product
- **SELF JOIN** – table joined to itself (e.g., employee-manager hierarchy)

**Q: Window functions — explain with example**
Operate across a set of rows related to the current row without collapsing them (unlike GROUP BY).
```sql
SELECT
  employee_id,
  department,
  salary,
  RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS dept_rank,
  SUM(salary) OVER (PARTITION BY department) AS dept_total,
  LAG(salary) OVER (PARTITION BY department ORDER BY hire_date) AS prev_salary
FROM employees;
```
Common functions: `ROW_NUMBER()`, `RANK()`, `DENSE_RANK()`, `LAG()`/`LEAD()`, `SUM()/AVG() OVER()`, `NTILE()`.

**Q: ROW_NUMBER vs RANK vs DENSE_RANK?**
- `ROW_NUMBER()` – unique sequential number, no ties.
- `RANK()` – ties get the same rank, but next rank skips (1,1,3).
- `DENSE_RANK()` – ties get the same rank, no gap (1,1,2).

**Q: Query optimization — what levers do you pull?**
- Check execution plan for table scans vs index seeks.
- Add appropriate indexes (careful: too many hurt write performance).
- Avoid `SELECT *`; select only needed columns.
- Avoid functions on indexed columns in `WHERE` (kills index usage — non-sargable).
- Filter early, reduce rows before joins.
- Update statistics; watch for parameter sniffing on stored procs.
- For big analytical tables, consider partitioning and columnstore indexes.

**Q: How would you find the second-highest salary?**
```sql
SELECT MAX(salary) FROM employees
WHERE salary < (SELECT MAX(salary) FROM employees);
-- or
SELECT salary FROM (
  SELECT salary, DENSE_RANK() OVER (ORDER BY salary DESC) AS rnk FROM employees
) t WHERE rnk = 2;
```

#### 6. ETL / ELT Concepts

**Q: ETL vs ELT?**
- ETL: transform before loading into the target (traditional, transform happens in an intermediate engine).
- ELT: load raw data first, transform inside the target system (common in cloud/lakehouse — leverages the scale of Synapse/Databricks/Snowflake compute).

**Q: Slowly Changing Dimensions (SCD) — explain Type 1 vs Type 2**
- **Type 1**: overwrite old value, no history kept (simple update).
- **Type 2**: preserve history — insert a new row with new surrogate key, mark old row as inactive (`is_current = 0`, `end_date`), start a new record with `is_current = 1`, `start_date`. Requires surrogate keys and effective-dated columns.

**Q: How do you handle late-arriving data / duplicates?**

Watermarking with buffer windows, dedup using `ROW_NUMBER()` partitioned by natural key ordered by load timestamp (keep rank=1), idempotent MERGE/upsert logic instead of blind inserts.

**Q: What is partitioning and why does it matter?**

Splitting large tables/files by a column (commonly date) so queries and loads only touch relevant partitions — improves performance and reduces cost by scanning less data (partition pruning).

**Q: How do you ensure data quality in a pipeline?**

Schema validation on ingest, row count reconciliation between source and target, null/duplicate checks, referential integrity checks, automated alerts on threshold breaches, and a data quality/audit log table.
