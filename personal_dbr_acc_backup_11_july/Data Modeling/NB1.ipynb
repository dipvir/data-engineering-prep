### What is Data Modeling?

**Data Modeling** is the process of designing the logical and physical structure of a database or data warehouse to determine how data is stored, organized, connected, and accessed.

In the context of Data Engineering, it involves translating complex business requirements into structured tables (such as Facts and Dimensions) and defining the relationships between them. The primary goal is to ensure data accuracy, eliminate unnecessary redundancy, and optimize query performance for downstream analytics and reporting.

---

### The Three Phases of Data Modeling (Great for your notes):

If an interviewer asks you to elaborate on how you approach it, you can break it down into these three standard levels:

1. **Conceptual Data Model (The Business View):** A high-level design that identifies the core business entities and their relationships without getting into technical details (e.g., "A *Customer* makes a *Purchase* of a *Product*").
2. **Logical Data Model (The Structural View):** Adds details to the conceptual model by defining specific data attributes, primary keys, and exact relationships (e.g., `Customer_ID` links to `Order_ID` in a 1-to-Many relationship).
3. **Physical Data Model (The Technical Implementation):** The actual database-specific design. This includes defining table schemas, data types, indexing strategies, partitioning, and storage formats (e.g., implementing a Star Schema in Delta Lake using `Parquet` files and configuring partition columns).

### Entity Modeling and Dimensional Modeling are the two foundational pillars of data modeling.

They represent the two entirely different ways we organize data depending on *what the database is being used for*.

#### 1. Entity Modeling (ER Modeling / Relational)

**Entity-Relationship (ER) modeling** is used to design databases for software applications that handle thousands of fast, everyday transactions (like placing an Amazon order or swiping a credit card). This is known as **OLTP (Online Transaction Processing)**.

* **The Primary Goal:** To eliminate data redundancy and ensure fast, secure **writes/updates**.
* **How it works:** It uses **Normalization** (usually up to 3rd Normal Form or 3NF). If a customer's address changes, you only want to update it in *one single place* to prevent errors. Therefore, data is broken down into many small, interconnected tables.
* **Core Concepts:** 
    * **Entities:** The real-world objects (e.g., `Customer`, `Order`, `Product`).
    * **Attributes:** The properties of those objects (e.g., `Customer_Name`, `Price`).
    * **Relationships:** How they connect (One-to-One, One-to-Many, Many-to-Many).

---
#### 2. Dimensional Modeling (Kimball Methodology)

**Dimensional modeling** is used to design Data Warehouses and Data Lakes for analytics, reporting, and Business Intelligence (BI). This is known as **OLAP (Online Analytical Processing)**.

* **The Primary Goal:** To optimize for incredibly fast **reads, aggregations, and complex queries** across millions of rows.
* **How it works:** It uses **Denormalization**. Instead of making an analyst join 15 different normalized tables to find "Total sales in Nagpur for July," Dimensional Modeling groups data into simplified, flattened structures (like a **Star Schema**).
* **Core Concepts:**
    * **Fact Tables:** The numerical measurements/metrics of a business event (e.g., `Sales_Amount`, `Quantity`, `Discount`).
    * **Dimension Tables:** The descriptive context of who, what, where, and when (e.g., `Dim_Customer`, `Dim_Time`, `Dim_Location`).

---

#### Quick Comparison Table (Great for Interviews)

| Feature | Entity Modeling (ER) | Dimensional Modeling |
| --- | --- | --- |
| **System Type** | **OLTP** (Transactional - e.g., App backend) | **OLAP** (Analytical - e.g., Data Warehouse) |
| **Primary Goal** | Fast **Writes**, Updates, and Deletes | Fast **Reads**, Search, and Aggregations |
| **Design Technique** | Normalization (3NF) | Denormalization (Star Schema) |
| **Data Redundancy** | Zero redundancy | Intentional redundancy to speed up queries |
| **Common Databases** | Azure SQL, PostgreSQL, MySQL | Azure Databricks, Snowflake, Redshift |

### Traditional OLTP databases Vs Snowflake Schema in OLAP use normalization techniques. So they follow the same data modeling approach?

---

#### ✅ Correct Answer:

**No, they follow ENTIRELY different data modeling paradigms**, even though both involve normalization:

| Aspect | OLTP (Traditional Databases) | Snowflake Schema (OLAP) |
| --- | --- | --- |
| **Modeling Paradigm** | **Entity Modeling (ER)** | **Dimensional Modeling** |
| **Scope of Normalization** | **FULL** - Entire database normalized (3NF) | **PARTIAL** - Only dimension tables normalized |
| **Fact Table** | No concept of Fact table (transactional tables) | Fact table remains **denormalized** (has all foreign keys) |
| **Purpose** | Eliminate redundancy for fast **writes/updates** | Balance query performance with storage efficiency |
| **Query Pattern** | Single record lookups (`WHERE order_id = 123`) | Aggregate millions of rows (`SUM(sales) GROUP BY region`) |

---

#### The Key Distinction:

**OLTP (Entity Modeling):**
- **Everything** is normalized across the entire database
- All tables follow normalization rules
- Example: `Customer` → `Address` → `City` → `State` → `Country` (all normalized)

**Snowflake Schema (Dimensional Modeling):**
- Only **dimension tables** are normalized
- **Fact table** stays denormalized (contains foreign keys to all dimensions)
- Example: `Fact_Sales` (denormalized) ← `Dim_Product` ← `Dim_Subcategory` ← `Dim_Category` (dimensions normalized)

---

#### Why This Matters:

Both use normalization, but:
- **OLTP normalizes for data integrity** across the entire operational system.
- **Snowflake Schema normalizes dimensions for storage efficiency** while keeping the fact table denormalized for analytical performance.

---

#### Interview Pro Tip:

If asked this question, emphasize:

*"While both involve normalization, OLTP uses **Entity Modeling** where the **entire database** is normalized for transactional integrity. Snowflake Schema uses **Dimensional Modeling** where only the **dimension tables** are normalized, while the **fact table remains denormalized** to maintain analytical query performance. They're fundamentally different approaches serving different purposes."*

### Why Interviewers Focus on Data Modeling

1. **Storage & Query Performance:** Poor data modeling leads to massive data duplication, high cloud storage costs, and extremely slow join queries across large Databricks/Data Warehouse clusters.
2. **Business Alignment:** A Data Engineer must know how to translate complex business requirements into clear **Fact and Dimension tables** or **Medallion layers (Bronze/Silver/Gold)**.
3. **Pipeline Efficiency:** How you model your target tables directly dictates how you handle incremental loads, Change Data Capture (CDC), and Slowly Changing Dimensions (SCD).

---

### Core Data Modeling Concepts You Must Master

#### 1. Dimensional Modeling (OLAP / Data Warehousing)

This is asked in almost every enterprise interview (Big 4, product firms, IT services):

* **Star Schema vs. Snowflake Schema:** 
    * **Star Schema:** Denormalized dimensions joined directly to a central Fact table (faster query performance, slightly more storage).
    * **Snowflake Schema:** Normalized dimensions (slower join performance due to multiple joins, but eliminates redundancy).

* **Fact Tables vs. Dimension Tables:**
    * **Fact Tables:** Contain numeric measures/metrics (e.g., `sales_amount`, `quantity_sold`) and foreign keys linking to dimensions.
    * **Dimension Tables:** Contain descriptive context (e.g., `customer_name`, `product_category`, `store_location`).
    * **Defining the Grain:** Understanding the lowest level of detail stored in a fact table (e.g., *one row per transaction* vs. *one row per daily summary*).

#### 2. Slowly Changing Dimensions (SCDs)

Interviewers frequently ask scenario questions on tracking history over time:

* **SCD Type 1:** Overwrite old values with new ones (no historical record).
* **SCD Type 2:** Insert a new row with active flags (`is_current`, `start_date`, `end_date`) to retain complete audit history.
* **Implementation:** Knowing how to execute SCD Type 2 using Delta Lake `MERGE INTO` in PySpark/SQL.

#### 3. Modern Lakehouse Modeling (Medallion Architecture)

Since you work with **Azure Databricks and Delta Lake**, you will definitely be asked about Lakehouse design patterns:

* **Bronze Layer (Raw):** Ingestion zone retaining raw source formats (JSON, CSV, streaming landing).
* **Silver Layer (Cleansed/Enriched):** Cleaned, deduplicated, and conformed data layer (often modeled in 3NF or flat structures).
* **Gold Layer (Curated/Business Ready):** Aggregated data modeled as Star Schemas or flattened analytical reporting tables optimized for BI tools and APIs.

#### 4. Physical Optimization & Storage Design

* **Partitioning vs. Bucketing:** When to partition by date/region vs. bucketing by high-cardinality keys.
* **Delta Lake Optimizations:** Understanding Liquid Clustering, `OPTIMIZE`, and `Z-ORDERING` to avoid over-partitioning and small file issues.

### HTAP (Hybrid Transactional/Analytical Processing)

**HTAP** represents the evolution of databases that can handle **BOTH** transactional (OLTP) and analytical (OLAP) workloads on the **same system** without requiring separate data warehouses.

---

### The Evolution of Database Storage Models

#### Traditional OLTP Era (Pre-2010s): Row-Based Storage Only

**Examples:** MySQL, PostgreSQL, older Oracle versions

**Storage Model:**
- Data stored **row-by-row** on disk
- Example: `| CustomerID: 123 | Name: John | City: Pune | Age: 30 |` stored contiguously

**Optimized For:**
- Transactional operations: `INSERT`, `UPDATE`, `DELETE` single records
- Point lookups: `SELECT * FROM customers WHERE customer_id = 123`

**Problem:**
- Analytical queries were extremely slow
- `SELECT AVG(age) FROM customers` (1M rows) → Had to read ALL columns from ALL rows, even though we only needed the `age` column

---

#### Modern HTAP Era (2010s onwards): Row-Based + Columnstore Indexes

**Examples:** SQL Server (2012+), Azure SQL Database, Oracle (In-Memory Column Store), PostgreSQL (with extensions)

**Storage Model:**
- **Primary data:** Still row-based (for OLTP performance)
- **Added capability:** Optional **columnstore indexes** for analytical queries

**How Columnstore Indexes Work:**

Instead of storing data row-by-row:
```
Row 1: | ID: 1 | Name: Alice | City: Mumbai | Sales: 5000 |
Row 2: | ID: 2 | Name: Bob | City: Delhi | Sales: 7000 |
Row 3: | ID: 3 | Name: Charlie | City: Pune | Sales: 6000 |
```

Columnstore stores **column-by-column**:
```
ID column: [1, 2, 3]
Name column: [Alice, Bob, Charlie]
City column: [Mumbai, Delhi, Pune]
Sales column: [5000, 7000, 6000]
```

**Benefits:**
- **10x-100x faster** for analytical queries
- `SELECT AVG(sales) FROM transactions` → Only reads the `sales` column, skips all other columns
- Better compression (similar values stored together compress better)

**Example in Azure SQL:**
```sql
-- Traditional row-based table (OLTP)
CREATE TABLE Orders (
    OrderID INT PRIMARY KEY,
    CustomerID INT,
    OrderDate DATE,
    Amount DECIMAL(10,2)
);

-- Add a columnstore index for analytical queries (HTAP capability)
CREATE COLUMNSTORE INDEX IX_Orders_Columnstore ON Orders;
```

**Key Distinction - HTAP Systems:**
- **OLTP operations** (INSERT, UPDATE, DELETE single orders) → Use row-based storage
- **OLAP operations** (SUM, AVG, GROUP BY across millions of orders) → Use columnstore index
- **Same database, both workloads!**

---

#### Pure OLAP Era (Cloud Data Warehouses): Columnar Storage by Default

**Examples:** Databricks (Delta Lake), Snowflake, Amazon Redshift, Google BigQuery

**Storage Model:**
- **Entire tables** stored in columnar format by default
- Built from the ground up for analytics
- Not optimized for high-frequency single-row updates

**Storage Format:**
- Databricks uses **Parquet** (columnar format) in Delta Lake
- Automatically compresses and optimizes for scan performance

---

### Quick Classification Guide

| Database Type | Storage Model | Columnstore Support | Best For |
| --- | --- | --- | --- |
| **Traditional OLTP** (older MySQL, PostgreSQL) | Row-based only | ❌ No | App backends, high-volume writes |
| **Modern HTAP** (SQL Server, Azure SQL, Oracle) | Row-based + Optional columnstore | ✅ Yes (added ~2012) | Mixed workloads (transactions + some analytics) |
| **Pure OLAP** (Databricks, Snowflake, Redshift) | Columnar by default | ✅ Yes (native) | Large-scale analytics, BI, ML |

---

### Real-World Architecture Example:

**Typical Enterprise Data Flow:**

1. **Azure SQL Database (HTAP - Primary OLTP):**
   - Handles your application's transactions (orders, user logins, inventory updates)
   - Can run *some* analytical queries via columnstore indexes
   - But limited scale for heavy analytics

2. **Azure Data Factory (ETL/ELT):**
   - Extracts data from Azure SQL
   - Loads into Databricks Lakehouse

3. **Azure Databricks (Pure OLAP):**
   - Stores historical data in Delta Lake (Parquet columnar format)
   - Runs massive analytical queries across billions of rows
   - Powers BI dashboards, ML models, and reports

### 🔥 Why Does Parquet Achieve 10-15x Better Compression Than CSV?

**Question:**

*"You mentioned that Databricks uses Parquet format in Delta Lake. Why does Parquet achieve such high compression ratios (10-15x) compared to CSV or JSON? What's the technical reason behind this?"*

---

#### ✅ Strong Answer:

**"Parquet achieves superior compression because it's a columnar storage format, which means data is stored column-by-column rather than row-by-row. This allows similar values of the same data type to be stored together sequentially, creating patterns that compression algorithms can exploit efficiently."**

---

### The Technical Breakdown:

#### 1. Why Columnar Storage Enables Better Compression

**Row-based storage (CSV/JSON):**
```
Row 1: [1, "Alice", "Mumbai", 5000]
Row 2: [2, "Bob", "Delhi", 7000]
Row 3: [3, "Charlie", "Pune", 6000]
```
- Data types constantly switch: integer → string → string → integer
- No repeating patterns
- Hard to compress efficiently

**Columnar storage (Parquet):**
```
ID column:    [1, 2, 3]              ← All integers, sequential
Name column:  ["Alice", "Bob", "Charlie"] ← All strings, similar length
City column:  ["Mumbai", "Delhi", "Pune"]  ← Repeated values in large datasets
Sales column: [5000, 7000, 6000]    ← All integers, similar magnitude
```
- Each column has **homogeneous data types**
- **Repeating patterns** and **similar values** are grouped together
- Perfect for compression algorithms

---

#### 2. Column-Specific Compression Techniques Parquet Uses

Parquet doesn't just use generic compression - it intelligently applies **encoding schemes per column**:

**A. Dictionary Encoding** (for low-cardinality columns with repeated values)

If your `City` column has only 50 unique cities across 10 million rows:
```
Instead of storing:
["Mumbai", "Mumbai", "Delhi", "Mumbai", "Pune", "Mumbai", ...]

Parquet stores:
Dictionary: {0: "Mumbai", 1: "Delhi", 2: "Pune"}
Column data: [0, 0, 1, 0, 2, 0, ...]  ← Much smaller (integers instead of strings)!
```

**B. Run-Length Encoding (RLE)** (for consecutive repeated values)

If you have a sorted dataset:
```
Status column: ["Active", "Active", "Active", "Active", "Inactive", "Inactive"]

Encoded as:
["Active" × 4, "Inactive" × 2]
```

**C. Bit Packing** (for small integer ranges)

If your `Age` column ranges from 0-127, Parquet only uses **7 bits** per value instead of 32 bits (standard integer storage).

**D. Delta Encoding** (for incremental data)

For timestamps or sequential IDs:
```
Instead of: [1000, 1001, 1002, 1003]
Store:      [1000, +1, +1, +1]  ← Much smaller deltas
```

---

#### 3. Real-World Compression Impact

| Storage Format | Typical Compression Ratio | Example: 1 TB Dataset |
| --- | --- | --- |
| **CSV (uncompressed)** | 1x (baseline) | 1 TB |
| **CSV (gzipped)** | 3-5x | ~250 GB |
| **Parquet (columnar + encoding)** | **10-15x** | **~70-100 GB** |

---

### Why This Matters for Databricks Data Engineers:

#### 1. Lower Cloud Storage Costs
```
Scenario: 10 TB of raw CSV data ingested daily

- CSV in S3/ADLS: 10 TB × 30 days = 300 TB/month
- Parquet in Delta Lake: ~25 TB/month (12x compression)

Storage savings: ~92% reduction in storage costs!
```

#### 2. Faster Query Performance
```
Query: SELECT AVG(sales_amount) FROM transactions WHERE region = 'Mumbai'

- CSV: Must scan ALL columns of ALL rows (even though you only need sales_amount and region)
- Parquet: Only reads the sales_amount and region columns (columnar projection)
  + Dictionary lookup for "Mumbai" (ultra-fast filtering)
  
Query time: 10-100x faster with Parquet
```

#### 3. Optimized Data Pipeline Throughput
- Less data to transfer from S3/ADLS to Spark workers
- Less memory required to process data
- More efficient caching in Delta Lake

### Interview Follow-Up Questions You Might Get:

**Q: "What about write performance? Doesn't columnar storage slow down writes?"**

- Yes, columnar formats like Parquet have slightly slower write performance compared to row-based formats because they need to reorganize data by column and apply compression. However, in Data Engineering pipelines (especially Medallion Architecture), we optimize for read-heavy analytical workloads. In Delta Lake, we also use techniques like:
  - *Write-optimized row groups in Parquet*
  - *Delta Lake's transaction log for fast writes*
  - *`OPTIMIZE` and `Z-ORDER` commands to reorganize data post-ingestion

**So we accept the minor write overhead for massive read performance gains.**

---

**Q: "When would you NOT use Parquet?"**

- For streaming ingestion with very high-frequency, small writes (like IoT sensor data arriving every millisecond), I might use Delta Lake's optimized write path or land data in JSON/Avro first in the Bronze layer, then convert to Parquet in the Silver layer during batch processing. Parquet is not ideal for sub-second latency requirements or single-row updates.

---

### Power Phrase for Interviews:

- Parquet's columnar storage groups similar data types together sequentially, enabling column-specific encoding techniques like dictionary encoding, run-length encoding, and bit packing. This results in 10-15x compression ratios, which directly translates to lower cloud storage costs and faster query performance because Spark only needs to scan the specific columns required by a query, not entire rows. That's why Databricks uses Parquet as the underlying format for Delta Lake.
