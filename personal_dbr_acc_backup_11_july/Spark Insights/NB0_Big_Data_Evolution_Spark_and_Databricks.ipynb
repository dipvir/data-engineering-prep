### Evolution of Big Data Processing: From Hadoop to Databricks

A comprehensive overview of how the data processing landscape evolved from early distributed computing challenges to modern unified analytics platforms.

---

#### 1. The Rise of Big Data (Early 2000s)

Big Data emerged as organizations faced unprecedented data challenges characterized by the **3 Vs**:

* **Volume:** Exponential data growth reaching terabytes and petabytes
  * Sources: Web applications, mobile devices, IoT sensors, social media
  * Traditional databases couldn't scale economically

* **Variety:** Diverse data formats requiring flexible storage
  * Structured: Relational tables, CSV files
  * Semi-structured: JSON, XML, logs
  * Unstructured: Images, videos, text documents

* **Velocity:** Real-time processing demands
  * Streaming data from multiple sources
  * Need for immediate insights to drive business decisions

---

#### 2. The Hadoop Era (2004-2013)

**Origins:**
* **2004:** Google published groundbreaking papers on **Google File System (GFS)** and **MapReduce**
  * Demonstrated distributed storage and parallel processing at scale
* **2006:** *Doug Cutting* created **Apache Hadoop**, inspired by Google's work
  * Open-source implementation of distributed computing principles

**Key Components:**
* **HDFS (Hadoop Distributed File System):** Fault-tolerant distributed storage
* **MapReduce:** Programming model for parallel batch processing
* **Ecosystem Growth:** *Hive* (SQL on Hadoop), *Pig* (data flow scripting), *HBase*, *Oozie*

**Limitations:**
* Primarily batch-oriented processing (high latency)
* Complex programming model (verbose Java code)
* Disk-based operations led to slow performance
* Difficult to use for iterative algorithms (machine learning)

---

#### 3. The Spark Revolution (2009-2013)

**Development:**
* **2009:** *Matei Zaharia* developed **Apache Spark** at *UC Berkeley's AMPLab*
  * Research project addressing Hadoop's limitations
* **2010:** Open-sourced to the community
* **2013:** Donated to Apache Software Foundation, became top-level project

**Key Innovations:**
* **In-Memory Processing:** Up to 100x faster than MapReduce for iterative workloads
* **Unified Engine:** Batch processing, streaming, machine learning, graph processing
* **Easier APIs:** High-level APIs in Python, Scala, Java, R, and SQL
* **Resilient Distributed Datasets (RDDs):** Fault-tolerant, distributed data structures
* **Lazy Evaluation & DAG Optimization:** Intelligent query planning

---

#### 4. Databricks & Modern Data Platforms (2013-Present)

**Foundation:**
* **2013:** *Databricks* founded by the original creators of Spark
  * *Matei Zaharia*, *Ali Ghodsi*, *Reynold Xin*, and other AMPLab researchers
  * Mission: Make big data simple and accessible

**Key Contributions:**
* **Unified Analytics Platform:** Collaborative environment for data engineers, data scientists, and analysts
* **Delta Lake:** Open-source storage layer bringing ACID transactions to data lakes
* **MLflow:** Open-source platform for the complete machine learning lifecycle
* **Lakehouse Architecture:** Combines data warehouse performance with data lake flexibility
* **AI Integration:** Native support for generative AI, LLMs, and AI-driven analytics

**Industry Leadership:**
* Cloud-native platform (AWS, Azure, GCP)
* Enterprise-grade security, governance, and compliance
* Industry standard for AI/ML workloads and modern data engineering
* Continuous innovation in data and AI technologies

---

#### Key Takeaways for Interviews

1. **Hadoop solved Volume** but struggled with speed and complexity
2. **Spark solved Performance** with in-memory computing and unified APIs
3. **Databricks solved Usability** by providing a collaborative, managed platform
4. **Lakehouse Architecture** represents the convergence of data lakes and warehouses
5. Modern platforms emphasize: **Performance + Simplicity + Governance + AI Integration**

### What is Apache Spark?

#### Definition:
Apache Spark is an open-source, distributed computing engine for large-scale data processing.

---

#### Key Features:

* **In-Memory Processing:** Processes data in RAM → up to 100x faster than Hadoop MapReduce
* **Unified Engine:** Single platform for:
  * Batch processing
  * Real-time streaming
  * Machine learning (MLlib)
  * Graph processing (GraphX)
  * SQL queries (Spark SQL)

* **Easy-to-Use APIs:** Python (PySpark), Scala, Java, R, SQL
* **Fault Tolerance:** Automatically recovers from failures via RDDs (Resilient Distributed Datasets)
* **Lazy Evaluation:** Optimizes query plans before execution using DAG (Directed Acyclic Graph)

---

#### Core Components:

1. **Spark Core:** Foundation for distributed task scheduling and memory management
2. **Spark SQL:** Structured data processing with DataFrames
3. **Spark Streaming:** Real-time data processing
4. **MLlib:** Scalable machine learning library
5. **GraphX:** Graph computation

---

#### Why Spark?

* ✅ **Fast:** In-memory caching and optimized execution
* ✅ **Flexible:** Works with multiple data sources (HDFS, S3, Kafka, databases)
* ✅ **Scalable:** Runs on clusters from 1 to 1000+ nodes
* ✅ **Iterative Workloads:** Ideal for machine learning and graph algorithms

---

**Created:** 2009 at UC Berkeley AMPLab by Matei Zaharia  
**Open-Sourced:** 2010  
**Apache Top-Level Project:** 2013

### Apache Spark Architecture

Spark follows a Master-Worker (Driver-Executor) architecture for distributed computing.

---

#### Core Components:

**1. Driver Program**
* Runs the main() function and creates SparkContext
* Converts user code into tasks (DAG of stages)
* Schedules tasks on executors
* Coordinates and monitors execution
* Holds metadata about the application

**2. Cluster Manager**
* Allocates resources across applications
* Types:
  * **Standalone:** Spark's built-in cluster manager
  * **YARN:** Hadoop's resource manager
  * **Mesos:** General-purpose cluster manager
  * **Kubernetes:** Container orchestration platform

**3. Executors**
* Worker processes running on cluster nodes
* Execute tasks assigned by the driver
* Store computation results in memory or disk
* Return results to the driver
* Each executor has a fixed number of cores and memory

**4. Tasks**
* Smallest unit of work
* Sent to one executor
* Operates on a single data partition

---

#### Execution Flow:

```
[User Code] → [Driver] → [SparkContext] → [Cluster Manager]
                                              ↓
                                         [Executors]
                                         Task Task Task
                                              ↓
                                        [Results back to Driver]
```

**Step-by-Step:**
1. **Submit Application:** User submits Spark application
2. **Driver Launch:** Driver program starts and creates SparkContext
3. **Resource Request:** SparkContext connects to cluster manager to request resources
4. **Executor Allocation:** Cluster manager allocates executors on worker nodes
5. **Code Distribution:** Driver sends application code to executors
6. **Task Creation:** Driver converts code into DAG → stages → tasks
7. **Task Execution:** Driver schedules and sends tasks to executors
8. **Computation:** Executors execute tasks in parallel on data partitions
9. **Result Return:** Executors send results back to driver
10. **Completion:** Driver releases resources after completion

---

#### Key Concepts:

* **DAG (Directed Acyclic Graph):** Logical execution plan with no cycles
* **Stage:** Set of tasks that can run in parallel without data shuffling
* **Shuffle:** Data redistribution across partitions (expensive operation)
* **Partition:** Logical chunk of data processed by a single task
* **Lazy Evaluation:** Transformations are not executed until an action is called
* **Lineage:** Record of transformations to rebuild lost data (fault tolerance)

---

#### Data Abstractions:

1. **RDD (Resilient Distributed Dataset):** Low-level, immutable, distributed collection
2. **DataFrame:** Structured data with named columns (like SQL table)
3. **Dataset:** Type-safe DataFrame (Scala/Java)

---

#### Advantages of This Architecture:

* **Scalability:** Add more executors for larger workloads
* **Fault Tolerance:** Re-execute failed tasks using lineage
* **In-Memory Processing:** Cache data in executor memory for speed
* **Parallel Processing:** Multiple tasks execute simultaneously
* **Resource Isolation:** Each application gets dedicated executors


### Spark Deployment Modes

Apache Spark can run in different deployment modes depending on where the driver program executes and how the cluster resources are managed.

---

#### 1. Cluster Mode

**How it works:**
* Driver runs **inside the cluster** on a worker node
* Cluster manager launches the driver as part of the application
* Entire application runs on the cluster

**Characteristics:**
* ✅ Driver failures can be recovered by the cluster manager
* ✅ Best for production workloads
* ✅ No need for client machine to stay connected
* ✅ Network overhead is minimized (driver is close to executors)
* ❌ Less convenient for interactive development

**Use cases:**
* Production jobs
* Scheduled/automated pipelines
* Long-running applications

---

#### 2. Client Mode

**How it works:**
* Driver runs **on the client machine** (outside the cluster)
* Executors run on cluster worker nodes
* Client machine must remain connected throughout execution

**Characteristics:**
* ✅ Interactive and easier to debug
* ✅ Good for development and testing
* ✅ Direct access to driver logs and output
* ❌ Client machine must stay connected
* ❌ Network latency between driver and executors
* ❌ Client becomes a bottleneck for large result collections

**Use cases:**
* Interactive notebooks (Databricks notebooks use this internally)
* Development and debugging
* Ad-hoc analysis
* spark-submit from local machine

---

#### 3. Local Mode

**How it works:**
* Both driver and executors run **on a single machine** (your local machine)
* Simulates distributed processing with threads
* No cluster manager needed

**Variations:**
* `local` - 1 thread
* `local[N]` - N threads
* `local[*]` - as many threads as logical cores

**Characteristics:**
* ✅ Easy setup - no cluster required
* ✅ Great for learning, testing, and small datasets
* ✅ Fast iteration during development
* ❌ Limited by single machine resources
* ❌ Not suitable for production or big data

**Use cases:**
* Learning Spark
* Unit testing
* Small-scale prototyping
* Local development

---

#### Deployment Mode vs Cluster Manager

**Deployment Mode** (where driver runs):
* Cluster mode
* Client mode
* Local mode

**Cluster Manager** (resource allocation system):
* **Standalone:** Spark's built-in cluster manager
* **YARN:** Hadoop's resource manager (supports both cluster and client modes)
* **Mesos:** General-purpose cluster manager
* **Kubernetes:** Container orchestration (supports both cluster and client modes)

---

#### Comparison Table

| Feature | Local Mode | Client Mode | Cluster Mode |
|---------|-----------|-------------|-------------|
| **Driver Location** | Local machine | Client machine | Cluster node |
| **Executors Location** | Local machine (threads) | Cluster | Cluster |
| **Cluster Required** | No | Yes | Yes |
| **Client Connection** | N/A | Must stay connected | Can disconnect |
| **Production Ready** | No | Not recommended | Yes |
| **Interactive** | Yes | Yes | Limited |
| **Fault Tolerance** | None | Limited | High |

---

#### Databricks Context

In **Databricks**:
* Notebooks and interactive workloads use a **managed client-mode-like architecture**
* The driver runs on the cluster's driver node (not your laptop)
* Jobs can be scheduled and run in a **cluster-mode-like fashion**
* You don't need to explicitly manage deployment modes
* Databricks handles driver placement, resource management, and fault tolerance automatically

---

**Key Takeaway:**
* **Local mode:** Development and testing on your machine
* **Client mode:** Interactive work, driver on client
* **Cluster mode:** Production jobs, driver on cluster


### Databricks Compute Types (Latest Releases)

The platform offers two major compute tiers: **Serverless** and **Classic**.

---

#### 1. Serverless Compute (Recommended)

Fully managed compute where Databricks handles all infrastructure — no cluster configuration, no instance types, no startup wait time.

**There are 3 types of serverless compute:**

* **Serverless compute for notebooks:** Interactive development with near-zero startup latency. Available to all workspace users with no additional permissions. Supports Python, SQL, and shell (R/Scala not supported on serverless).
* **Serverless compute for jobs:** Run jobs without configuring infrastructure. Autoscaling and Photon are automatically enabled. Databricks optimizes and scales compute for your workload.
* **Serverless compute for pipelines:** Used by Lakeflow Spark Declarative Pipelines (SDP). Enhanced autoscaling (both horizontal and vertical), stream pipelining, and incremental materialized view refreshes.

> **Note:** Serverless SQL warehouses (covered in Section 5) are a separate product that *runs on* serverless infrastructure, but are not classified as "serverless compute." Other features that also run on serverless infrastructure include model training forecasting, data quality monitoring, and predictive optimization.

**Key advantages:**
* Near-zero startup latency
* No infrastructure management (no cluster policies, no instance types)
* No cluster creation permission required
* Always uses Unity Catalog
* Pay only for what you use
* Automatic autoscaling and Photon acceleration

---

#### 2. Classic Compute (Traditional Clusters)

You manually configure clusters including instance types, autoscaling, and cluster policies. Uses Databricks Runtime (DBR).

**2a. All-Purpose Clusters (Interactive)**
* Used for interactive development, exploration, and ad-hoc analysis
* Multiple users can share a single cluster (attach multiple notebooks)
* Can be started/stopped manually or auto-terminated after inactivity
* More expensive per DBU than jobs clusters
* Best for: Notebooks, EDA, debugging, collaborative development

**2b. Jobs Clusters (Automated)**
* Dedicated to running a single job, then terminated automatically
* Lower cost per DBU (jobs pricing tier)
* Cannot be shared across multiple users
* Created and managed by the job scheduler
* Best for: Scheduled pipelines, ETL jobs, automated workflows, production workloads

---

#### 3. Access Modes (Security Layer)

Every compute resource has an access mode that determines who can use it and what data they can access.

| Access Mode | Former Name | Description | Supported Languages |
|---|---|---|---|
| **Standard** | Shared | Multi-user with data isolation via Databricks Lakeguard. Recommended for most workloads. | Python, SQL, Scala |
| **Dedicated** | Single User | Assigned to one specific user or group for exclusive access. | Python, SQL, Scala, R |
| **Legacy: No Isolation Shared** | No Isolation | Deprecated, not recommended. No data isolation between users. | Python, SQL, Scala, R |

* Access mode defaults to **Auto** — the platform selects Standard automatically unless you choose a ML runtime, GPU instance, or DBR below 14.3 (which defaults to Dedicated).
* Databricks recommends **Standard** access mode unless specific functionality is not supported.

---

#### 4. Cluster Configurations

**Single-Node Compute:**
* Runs Spark locally — driver acts as both master and worker
* No worker nodes, one executor thread per logical core minus 1
* Best for small datasets, non-distributed ML libraries, and cost savings
* GPU scheduling is NOT enabled on single-node compute
* Cannot be converted to multi-node

**Multi-Node Compute:**
* Driver + worker nodes for distributed workloads
* Can autoscale from min to max workers
* Cannot scale to 0 workers
* Best for large-scale data processing, ETL, distributed ML

**GPU-Enabled Compute:**
* Requires Machine Learning runtime (GPU ML version auto-selected)
* Photon is NOT supported with GPU instances
* Supported GPU types include A100 (all clouds), and various NVIDIA types
* Can be single-node (single GPU) or multi-node (distributed GPU)
* Best for deep learning, LLM fine-tuning, GPU-intensive computation

---

#### 5. SQL Warehouses

* Compute resources optimized for running SQL queries (Databricks SQL)
* Available as **Serverless** (instant start, managed) or **Classic** (configured)
* Used by: SQL editor, dashboards, alerts, BI tools, Genie spaces
* Separate from all-purpose/jobs clusters — dedicated to SQL workloads

---

#### Comparison Summary

| Feature | Serverless | Classic All-Purpose | Classic Jobs | SQL Warehouse |
|---|---|---|---|---|
| **Startup Time** | Near-zero | Minutes | Minutes | Near-zero (Serverless) |
| **Infrastructure Management** | Databricks manages | You configure | You configure | Databricks manages (Serverless) |
| **Cost per DBU** | Serverless pricing | Higher | Lower | Serverless/Classic pricing |
| **Use Case** | Notebooks, jobs, pipelines | Interactive dev | Automated jobs | SQL queries, BI |
| **Autoscaling** | Automatic (H + V) | Manual config | Manual config | Automatic |
| **Photon** | Auto-enabled | Optional | Optional | Auto-enabled (Serverless) |
| **Cluster Creation Permission** | Not required | Required | Required | Not required (Serverless) |

---

#### Key Takeaways

* **Serverless** is the recommended default — no setup, fast startup, lower overhead
* Use **Jobs compute** instead of all-purpose for any non-interactive workload (lower cost)
* **Standard** access mode is recommended for multi-user workloads with Unity Catalog
* **Single-node** compute saves cost for small workloads, use **multi-node** for distributed processing
* **GPU compute** requires ML runtime and is essential for deep learning workloads
* Databricks abstracts away cluster manager selection (no YARN/Mesos/K8s to configure)
