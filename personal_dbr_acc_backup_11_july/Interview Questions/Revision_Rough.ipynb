```sql
SELECT
    * 
FROM 
    (SELECT user_id, name, email, signup_date, row_number() over (partition by name, email order by signup_date) as rank FROM users) AS ranked
WHERE rank > 1
```

```sql
DELETE FROM users 
WHERE user_id in (SELECT
                        user_id
                    FROM 
                        (SELECT user_id, name, email, signup_date, row_number() over (partition by name, email order by signup_date) as rank FROM users) AS ranked
                    WHERE rank > 1
                )
```

```sql
%python

n = int(input("Enter a number: "))

display(
spark.sql(
    f"""
    SELECT
        *
    FROM
        (SELECT 
            *,
            dense_rank() OVER (ORDER BY salary DESC) as rank
        FROM employees) AS ranked
    WHERE rank = :n
    """, args={"n" : n}
))
```

```sql
SELECT
  *,
  row_number() OVER (ORDER BY salary DESC) as row_number,
  rank() OVER (ORDER BY salary DESC) as rank,
  dense_rank() OVER (ORDER BY salary DESC) as dense_rank
FROM
  employees
```
